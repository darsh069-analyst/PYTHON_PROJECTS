- Temperature Analysis

Generate random data for 30 days and calculate the mean, median, and standard deviation of the temperature over the month. also, find the days with temperature above the average

```python
import numpy as np
weather = np.random.uniform(22, 40, 30)
weather = np.round(weather, 2)
print(weather)
```

```python
#Mean Temperature

Mean_weather = np.round(np.mean(weather), 2)
print("Mean weather:", Mean_weather)
```

```python
#Median Temperature

Median_weather = np.round(np.median(weather), 2)
print("Median weather:", Median_weather)
```

```python
#Standard Deviation

Standard_Deviation_weather = np.round(np.std(weather), 2)
print("Standard Deviation weather:", Standard_Deviation_weather)
```

```python
#Days with temperature above the average
weather[weather > Mean_weather]
```
