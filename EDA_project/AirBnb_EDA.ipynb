### 1: Importing all Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

%matplotlib inline
```

### 2: Loading the Dataset

```python
# Reading the dataset from a CSV file into a pandas DataFrame
dataset = pd.read_csv('datasets.csv' , encoding_errors = 'ignore')
```

### 3. Initial Exploration

```python
# Checking the information of the dataset
dataset.info()
```

```python
# Statistical summary of the dataset
dataset.describe()
```

### 4: Data Cleaning

```python
# Checking for Null values
dataset.isnull().sum()

# Drop rows with missing values
dataset.dropna(inplace=True)
```

```python
# Checking for duplicates
dataset.duplicated().sum()

#Deleting duplicates
dataset.drop_duplicates(inplace=True)
```

```python
# Checking the data types of each column
dataset.dtypes
```

```python
# Converting the 'id' column to string type
dataset['id'] = dataset['id'].astype(str)
```

## EDA
### 5: Data Analysis

#### ** Univariate Analysis **
##### (Single Variable analysis)

```python
# Identifying outliers in the 'price' column using boxplot

dataset = dataset[dataset['price'] < 1500]
sns.boxplot(data = dataset, x=dataset['price'])
```

```python
# Price distribution

plt.figure(figsize=(6,3))
sns.histplot(data = dataset, x="availability_365")
plt.title('Availability_365 Distribution')
plt.xlabel('Availability (365 days)')
plt.ylabel('Frequency')
plt.show()
```

#### Feature Engineering

```python
#['price per bed']

dataset['price per bed'] = dataset['price']/dataset['beds']
```

```python
# Price per bed
dataset.groupby(by='neighbourhood_group')['price per bed'].mean()
```

```python
# Price dependency on neighborhood
dataset.groupby(by='neighbourhood_group')['price'].mean()
```

#### Bi Variable Analysis
##### (One variable dependency in another variable)

```python
# Price dependency on neighborhood 
sns.barplot(data=dataset, x='neighbourhood_group', y='price', hue='room_type')
```

```python
# Number of reviews and price relationship
plt.figure(figsize=(8,5))
plt.title('Location wise price distribution')
sns.scatterplot(data=dataset, x='number_of_reviews', y='price', hue='neighbourhood_group')
```

```python
# Pairplot to visualize the relationships between multiple variables
sns.pairplot(data=dataset, vars=['price', 'minimum_nights', 'number_of_reviews', 'availability_365'], hue='room_type')
```

```python
# Geographical distribution of AirBnb listings
plt.figure(figsize=(10,7))
sns.scatterplot(data=dataset, x='longitude', y='latitude', hue='room_type')
plt.title('Geographical Distribution of AirBnb Listings')
plt.show()
```

```python
# Creating a heatmap - correlation of one variable with the relationships between numerical variables
corr = dataset[['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month','availability_365', 'beds' ]].corr()
corr

plt.figure(figsize=(8,6))
sns.heatmap(data = corr, annot=True)
```
